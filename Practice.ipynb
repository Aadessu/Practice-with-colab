## Functions

```python
def greet(name):
    return f"Hello, {name}!"


def add(first, second):
    return first + second


print(greet("Ada"))
print("5 + 3 =", add(5, 3))
```

## Loops

```python
for number in range(1, 6):
    print(number)

count = 3
while count > 0:
    print("Countdown:", count)
    count -= 1
```

## Conditionals

```python
temperature = 24

if temperature > 30:
    print("It is hot")
elif temperature >= 20:
    print("The weather is pleasant")
else:
    print("It is cold")
```

## Exception handling

```python
try:
    number = int("not a number")
except ValueError as error:
    print("Could not convert the value:", error)
finally:
    print("The conversion attempt is complete")
```

## Libraries

```python
import math
import random

print("Square root of 16:", math.sqrt(16))
print("Random number from 1 to 10:", random.randint(1, 10))
```
